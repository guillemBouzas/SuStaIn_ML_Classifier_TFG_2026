```python
# imports

import pandas as pd
import json
```

```python
df1 = pd.read_csv('nextbrain_L_vols.csv', sep = ',')
df2 = pd.read_csv('nextbrain_R_vols.csv', sep = ',')
df3 = pd.read_csv('other_info.csv', sep = ',')
```

```python
columnes = df2.columns.tolist()
dfNextbrain = df2.copy()

print(dfNextbrain.shape)
print(dfNextbrain['subject'].nunique())

for col in columnes:
    if col != 'subject':
        dfNextbrain[col] = df1[col] + df2[col] # El nou dataset tindrà els volums sumats

# Eliminem ja unes quantes columnes innecessàries
dfNextbrain = dfNextbrain.drop(columns = ['Unnamed: 0', 'session'])
```

```python
# comprovem el df
dfNextbrain.head()
```

```python
# Anem a crear  el df per les regions de l'hipocamp

# Obrim el diccionari
with open('AllenAtlasLUT_xtra.json', 'r') as f:
    lut_data = json.load(f)

# classes que pertanyen a l'hipocamp
CLASS_HI_IDS = [ 344, 369, 409, 567, 340, 563,
    419, 373, 572, 345, 370, 374,
    410, 420, 564, 573, 341, 568,
    371, 375, 411, 421, 565, 569,
    574, 342, 346, 326, 347, 576,
    558, 404, 364, 405, 559, 365,
    561, 407, 367, 575, 422, 432,
    320, 461, 343, 368, 372, 408, 418,
    562, 566, 571, 339, 354]

columnas_hi = []
for struct_id in CLASS_HI_IDS:
    str_id = str(struct_id)
    if str_id in lut_data:
        nombre_columna = lut_data[str_id]['labelName']
        if nombre_columna in dfNextbrain.columns:
            columnas_hi.append(nombre_columna)

if 'subject' in dfNextbrain.columns:
    df_hi = dfNextbrain[['subject'] + columnas_hi]
else:
    df_hi = dfNextbrain[columnas_hi]
```

```python
# Mirem algunes dades del df que té les dades de nextBrain només de la regió de l'hipocamp

print(df_hi.shape)

df_hi.head()
```

```python
# Fem l'agrupació de columnes de l'hipocampus per reduir-ne el nombre

# Llista dels subcamps CA que volem agrupar
ca_subfields = ['ca1', 'ca2', 'ca3', 'ca4']

# Identificar quines columnes del fitxer original contenen algun CA
ca_cols = [col for col in df_hi.columns if any(ca in col.lower() for ca in ca_subfields)]

# Identificar les columnes que NO contenen CA (les que volem deixar com estan)
other_cols = [col for col in df_hi.columns if col not in ca_cols]

# Creem el nou DataFrame començant per les columnes que no canvien
new_df = df_hi[other_cols].copy()

# Per a cada subcamp CA, sumem totes les seves capes i regions en una sola columna
for ca in ca_subfields:
    # Seleccionem només les columnes d'aquest CA específic (ex: tot el que tingui 'ca1')
    target_cols = [col for col in df_hi.columns if ca in col.lower()]

    if target_cols:
        new_df[ca.upper()] = df_hi[target_cols].sum(axis=1)

# Reordenem per a que 'subject' sigui la primera columna si no ho era ja
cols = ['subject'] + [c for c in new_df.columns if c != 'subject']
new_df = new_df[cols]
```

```python
# Anem a veure com ens queda finalment amb l'agrupament

print(new_df.shape)

new_df.head()
```

```python
# Anem ara a gestionar el df amb other info

df_relevant = df3[['abeta_pos', 'age', 'apoe4']]
```

```python
# Afegim les columnes al nostre df net

df_resultado = pd.concat([new_df, df_relevant], axis=1)

# Comprovem aquest primer preprocés:
print(df_resultado.shape)

df_resultado.head()

# Ens el guardem
df_resultado.to_csv('df_hp_adni.csv', index=False, sep=',', encoding='utf-8')
```
