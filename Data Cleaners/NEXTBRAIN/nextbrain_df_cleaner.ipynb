Primer obrirem els dos df del NextBrain i els ajuntarem en un de sol

```python
# imports
import pandas as pd
```

```python
df1 = pd.read_csv('nextbrain_L_vols.csv', sep = ',')
df2 = pd.read_csv('nextbrain_R_vols.csv', sep = ',')
```

```python
columnes = df2.columns.tolist()
dfNextbrain = df2.copy()

for col in columnes:
    if col != 'subject':
        dfNextbrain[col] = df1[col] + df2[col] # El nou dataset tindrà els volums sumats

# Eliminem ja unes quantes columnes innecessaries
dfNextbrain = dfNextbrain.drop(columns = ['Unnamed: 0', 'session'])

# Guardem el df net
(dfNextbrain.to_csv('nextbrain_vols.csv', index=False, sep=',', encoding='utf-8'))
```

Ara que tenim el df net anem a agrupar les parts de l'hipocamp

```python
import json

df = pd.read_csv('nextbrain_vols.csv')

with open('AllenAtlasLUT_xtra.json', 'r') as f:
    lut_data = json.load(f)


CLASS_HI_IDS = [ 344, 369, 409, 567, 340, 563,
    419, 373, 572, 345, 370, 374,
    410, 420, 564, 573, 341, 568,
    371, 375, 411, 421, 565, 569,
    574, 342, 346, 326, 347, 576,
    558, 404, 364, 405, 559, 365,
    561, 407, 367, 575, 422, 432,
    320, 461, 343, 368, 372, 408, 418,
    562, 566, 571, 339, 354]

columnas_hi = []
for struct_id in CLASS_HI_IDS:
    str_id = str(struct_id)
    if str_id in lut_data:
        nombre_columna = lut_data[str_id]['labelName']
        # Verifiquem si aquesta columna existeix realment en el DataFrame
        if nombre_columna in df.columns:
            columnas_hi.append(nombre_columna)

# Incloem 'subject'
if 'subject' in df.columns:
    df_hi = df[['subject'] + columnas_hi]
else:
    df_hi = df[columnas_hi]

# Resultato
print("Columnes trobades per CLASS_HI:", columnas_hi)
print(df_hi.head())
```

```python
print(df_hi.shape)

# Guardem el df net
(df_hi.to_csv('nextbrain_hippocampus_vols.csv', index=False, sep=',', encoding='utf-8'))
```
