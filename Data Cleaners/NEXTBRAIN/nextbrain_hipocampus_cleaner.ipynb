```python
import pandas as pd

# Carreguem el fitxer
df = pd.read_csv('nextbrain_hippocampus_vols.csv')
```

```python
df.shape
```

```python
# Llista dels subcamps CA que volem agrupar
ca_subfields = ['ca1', 'ca2', 'ca3', 'ca4']

# Identificar quines columnes del fitxer original contenen algun CA
ca_cols = [col for col in df.columns if any(ca in col.lower() for ca in ca_subfields)]

# Identificar les columnes que NO contenen CA (les que volem deixar com estan)
other_cols = [col for col in df.columns if col not in ca_cols]

# Creem el nou DataFrame començant per les columnes que no canvien
new_df = df[other_cols].copy()

# Per a cada subcamp CA, sumem totes les seves capes i regions en una sola columna
for ca in ca_subfields:
    # Seleccionem només les columnes d'aquest CA específic (ex: tot el que tingui 'ca1')
    target_cols = [col for col in df.columns if ca in col.lower()]

    if target_cols:
        new_df[ca.upper()] = df[target_cols].sum(axis=1)

# Reordenem per a que 'subject' sigui la primera columna si no ho era ja
cols = ['subject'] + [c for c in new_df.columns if c != 'subject']
new_df = new_df[cols]

# Guardar el resultat
new_df.to_csv('nextbrain_only_CA_grouped.csv', index=False)

print("Processament completat.")
print(f"Columnes mantingudes intactes: {len(other_cols) - 1}") # -1 per 'subject'
print(f"Noves columnes de CA creades: {[ca.upper() for ca in ca_subfields]}")
```
